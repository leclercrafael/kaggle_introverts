```python
import pandas as pd
```

Import of datasets

```python
train = pd.read_csv('data/train.csv')
test = pd.read_csv('data/test.csv')
index = test.id
```

```python
train = train.drop(columns=['id'])
test = test.drop(columns=['id'])
train = train.rename(columns={'Personality': 'label'})
```

```python
from autogluon.tabular import TabularPredictor

predictor = TabularPredictor(label='label', problem_type='multiclass', eval_metric='accuracy')

predictor.fit(
    train_data=train, 
    time_limit=5400,  # Limite de temps en secondes (facultatif)
    num_stack_levels=1,  # ➜ stacking niveau 2
    presets='best_quality',
     verbosity = 3 # Pour un training plus poussé
)
```

```python
predictor.leaderboard(silent=True)
```

```python
preds = predictor.predict(test, model= 'WeightedEnsemble_L2')
submission_df = pd.DataFrame({
    'id': index,  # ou autre identifiant
    'Personality': preds  # ou renommer si besoin
})
submission_df.to_csv('data/sub_autogluon2.csv', index=False)
```
